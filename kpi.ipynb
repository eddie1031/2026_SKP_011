# KPI

항만 운영 KPI를 설정할때, 크게 세가지 정도로 나누어 볼 수 있다. 얼마나 많이 처리했는가, 배가 얼마나 오래 머물렀는가, 한꺼번에 얼마나 몰렸는가 정도로 나누어볼수 있다.

- 처리량(Throughput): 얼마나 많이 처리했는가
- 체류시간(Dwell Time): 배가 얼마나 오래 머물렀는가
- 혼잡도(Congestion): 한꺼번에 얼마나 몰렸는가

'체류시간' 이라는 문제를 해결할때, 어떤사람은 체류시간을 입항부터 출항까지, 어떤사람은 접안부터 이안까지.

지표마다 이름, 계산식, 단위, 집계 기간, 데이터 출처, 제외조건 등을 기록해둡니다. 이 기록을 KPI정의서 라고 일컫습니다.




## KPI 정의서 작성요령

### 목표 설정
무엇을 개선하려는지에 대한 내용을 설정한다. 예시로는 컨테이너 처리 규모 확대 같은 것들이 있을 수 있다.

### 성과의 증거
목표가 달성되면 어떤 값이 달라질 수 있는지를 설정한다. 예시로는 월간 처리 물동량이 개선될수 있다.

### 판단과 행동
값이 목표보다 낮을때, 누가 무엇을 어떻게 검토해야하는가에 대한 내용이 있다.

### 측정 가능성
필요한 데이터가 실제로 존재하고 일정한 시점에 확보되는가?

## 정의서에 포함될수 있는 항목

1. 식별정보: KPI 정의서 이름, 지표 이름, 명칭
2. 목적: 연결된 조직의 목표, 활용할 의사결정(왜 측정하는가?)
3. 측정 대상: 대상 업무, 시설, 고객, 지역(어디까지 측정하는가?)
4. 측정 단위: TEU, 건, 원, 시간, %(숫자의 단위는 명확한가?)
5. 집계 기간: 일, 월, 분기, 기간의 시작과 끝(어느 날짜를 기준으로 하는가?)
6. 산식(계산식): 합계 또는 분자/분모의 계산 방법(다른 사람이 같은 값을 계산할 수 있는가?)
7. 포함/제외 기준: 상태, 유형, 취소,중복,시험 데이터를 처리해야한다(어떤 행을 계산에 넣을것인가?)
8. 데이터 출처: 시스템, DB, 파일(원천값을 추적할 수 있는가?)
9. 갱신/확정: 수집 주기, 발표 시점, 수정 허용기간(언제의 값이 공식 실적인가?)
10. 목표와 비교: 목표값, 지표값, 목표 설정 근거(무엇과 비교해 좋은 성과라 판단할 수 있는가?)
11. 책임과 관리: 지표 책임자, 데이터 관리자, 검토 주기(오류와 부진을 누가 처리할 것인가?)
12. 해석과 한계: 외부 영향, 데이터 제약, 함께 볼 지표(이 숫자만 보고 잘못 판단할 위험은 없는 것인가?)

## 검수방법

가장 효과적인 검수는 정의서만 받은 다른 담당자한테 같은 기간의 KPI를 계산시켜보는 것. 값이 다르면 무언가 잘못되었다고 볼 수 있음

- 같은 월, 같은 원천데이터로 계산했을때 값이 같은가?
- 경계 시각의 기록, 취소건, 수정건을 어떻게 처리할지 결정했는가?
- 비율지표의 분자와 분모가같은 대상/기간을 가리키는가?
- 0, 결측, 분모 0을 서로 다르게 처리하는가?
- 잠정 실적과 확정 실적을 구분하는가?
- 목표값의 기간과 근거가 있는가?
- 지표가 변했을때 담당자가 취할 수 있는 합리적인 조치가 있는가?


## 처리량(Throughput)

처리량은 정해진 기간동안 항만이 싣고 내린 화물의 양으로 정리할 수 있다. 공장을 예시로 두면 생산량이 있을수 있고, 항만이 얼마나 일했는지를 가장 직접적으로 보여주는 지표로 볼 수 있다. 항만 순위나 연간 실적 보도에 가장 먼저 보여지기도 한다.

처리량에 사용되는 단위는 일반적으로는 TEU입니다.

TEU = 10피트 (0.5), 20피트(1), 40(2), 45피트(2.5, 2.25)

처리량만으로는 좋은지 나쁜지 판단이 어렵다. 전체 가운데 한 부분이 차지하는 비율 구성비(share)를 사용하거나, 이전 기간과 비교한 변화율인 증감률(Growth Rate)을 사용해서 비교한다.

- 구성비(%) =  부분값 / 전체값 * 100
- 증감율(%) = (이번 기간 값 - 이전 기간 값) / 이전 기간 값 * 100

분모가 0이면 계산 불가능합니다. 부두별 통계를 살펴보면, 개장하기 전년도는 0으로 기록되어있다. 0은 처리실적이 없는것이 아니라, 아직 부두가 없었다를 의미합니다. 개장 첫해의 증감율을 계산하면 무한대가 나온다. 이때에는 증감율을 신규 개장으로 따로 표시해야한다.

## 체류시간(Dwell Time)
체류시간은 선박이 항만에 들어온 순간부터 떠날떄까지 머문 시간을 의미한다. 선박 체류시간이 길어지면 뒤따라 들어오는 배가 선석을 기다리게된다. 선사입장에서는 배가 항만에 묶여있는 시간이 비용이 된다. 체류시간은 항만 서비스 품질을 평가하는 핵심 지표(KPI)가 될 수 있다.

체류시간은 어떤 두 시각 사이를 재느냐에 따라 달라진다. 입항 부터 출항까지를 체류시간으로 보면, 묘박지 정박 시간이 들어갑니다. 접안부터 이안까를 체류시간으로 보면, 하역작업시간만 포함된다.

### 시간연산

pandas에서 시간연산을 하기 위해서는 문자열로 되어있는 시간을 시간자료형(Timestamp)로 바꿔줘야 한다. Timestamp끼리 연산을 하면 Timedelta가 나온다. '1days 03:30:00'(Timedelta) 이 값을 초(Seconds)로 변경가능하다. 이 값을 다시 3600으로 나누면 시간(Hour)을 구할 수 있다.

숫자형은 변경실패시 NaN(Not A Number)이 나오고, 시간자료형은 변경 실패시 NaT(Not A Time)로 변경됨, NaT도 isna() 함수로 집계 가능하다.

### 평균값과 중앙값

체류시간을 KPI로 보고할때, 여러 선박의 값을 하나로 요약해야한다. 가장 일반적으로 사용하는 값은 평균값(Mean)이다. 체류시간처럼 대부분은 하루 안팎인데 수리나 계류때문에 몇 주씩 머무는 배가 가끔 섞이는 데이터가 있으면, 평균값이 가려진다. 열척 중 아홉척이 20시간, 한척이 500시간이면 평균적으로 20시간 머물러요 라는 가정이 성립이 안될 수 있다.

이럴때에는 중앙값(Median)을 살펴보는 것이 합리적이다. 중앙값은 극단적인 값 몇개에 흔들리지 않기 때문에, 전형적인 선박의 체류시간을 나타내는데 적합하다.

평균은 항만 전체가 떠안고있는 부담을 나타낸다. 둘 사이의 차이가 크면 긴 꼬리가 있다는 신호이고, 상위 10%가 넘는 값인 90번째 백분위수를 함께 포함해서 살펴보면, 긴 체류가 얼마나 긴지도 같이 확인할 수 있다.

## 혼잡도(Congestion)

혼잡도는 한정된 시설에 얼마나 많은 이용량이 몰렸는지를 나타내는 지표이다. 항만은 하루 입항 척수가 같더라도 특정 시간대 몰리거나 여러 배가 동시에 오래 머물면 선석, 예선, 도선사가 부족해진다.

혼잡도는 하나의 정해진 공식이 있는 지표가 아니라, 무엇을 '혼잡'으로 볼 것인가를 여러 방식으로 정의하는 것이 바람직하다.

- 특정 시각에 항만 안에 머물고 있는 선박수(동시 체류수)
- 하루중 어느 시간대에 입항이 몰리는가(시간대별 입항 척수)

### 동시체류수

동시 체류수는 특정 어떤 순간에 항만 안에 배가 몇 척 있는가?를 계산한 지표이다. 입출항을 이벤트(Event)로 두고, 입항할때 증가, 출항할때 감소 하는 식으로 계산하고, 이러한 이벤트들을 시간순서로 정렬한 뒤 누적합을 구한다.

관측 시작 시점의 경계 효과를 주의해서 살펴보아야 한다. 데이터가 1월 1일부터 시작하면, 12월 31일에 들어와서 1월 1일에 머물고 있는 배는 입항기록이 없다. 출항하면 출항 기록만 있다.

관측 초기에는 동시 체류 수가 실제보다 적게 나올수 있기 때문에 긴 체류가 충분히 쌓인 구간 만큼의 내용을 KPI로 쓰는것이 권장된다.

- 동시 체류 수 = 시작 지점까지의 입항 누적합 - 끝 시점까지의 출항 누적합
- 동시 체류 수 = t까지의 입항 건수 누적 - t까지의 출항 건수 누적
