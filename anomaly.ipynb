# 이상징후 탐지

관제실 모니터에 오늘 부산항 입항척수가 40척이다. 평소에는 120척정도 들어왔는데 40척이 들어왔다. 아 이거 부산항에 뭔일 있구나.

태풍, 관제 시스템이 먹통이 되어서 기록이 누락되었거나. 평소와 다르다를 판단하려면, 평소가 무엇인지를 결정해야하고, 얼마나 달라야 이상한것인지를 결정할 수 있어야 한다.

데이터에서 평소와 다른 값을 찾아서 담장자에게 알리는 이상징후 탐지(Anomaly Detection)를 살펴보겠다.

# 이상징후란

이상치(Outlier) 다른 값들과 동떨어져있는 값을 가리키는 통계 용어이다. 이상징후는 평소의 패턴에서 벗어나서 확인이 필요한 신호를 의미하는데, 모든 이상치는 이상징후가 아니다.

20만톤급 컨테이너선의  총 톤수는 다른 배들과 다소 떨어져있는 값일 수 있지만, 그런 이상치로 구분될수 있는 값이 들어왔다고 해서 무조건적으로 이상징후로 받아들일 필요는 없다.

그런데, 평균 120척 들어왔는데 원래는 목요일에는 80척정도가 들어왔던 상황에서 갑자기 목요일에 120대가 들어오면 이상징후로 볼 수 있다.

이상징후가 발견이 될때 원인을 구분해보면 크게 세가지로 구분해볼 수 있다.

- 데이터오류: 3200년 출항이 갑자기 잡힘, 환적 톤수가 20억, 중복기록
- 운영사건: 장비 고장, 환경문제, 명절
- 구조변화: 신항 개장, 코로나, 선사 파산

## 이상(異常)의 모양

이상징후는 모양에따라 세가지로 구분해볼 수 있다.

- 점 이상: 값 하나가 전체에서 동떨어진 경우
- 맥락 이상: 값 자체는 흔한데, 그 상황에서는 이상한 것
- 집단 이상: 값 하나하는 평범한데, 여러개가 모인 모양이 드문것
